```python
%%writefile program.l
%{
#include <stdio.h>
%}

%%

"//".*                                      ;   /* Remove single-line comments */

"/*"([^*]|\n|(\*+([^*/]|\n)))*"*"+"/"       ;   /* Remove multi-line comments */

[ \t\n]+                                    ;   /* Remove whitespace */

.                                           { printf("%s", yytext); }

%%

int yywrap()
{
    return 1;
}

int main()
{
    yylex();
    return 0;
}
```

if you want to use at txt as an input

```python
%%writefile program.txt
#include<stdio.h>

// This is a comment

int main()
{
   /* Multi-line
      comment */
   int a = 10;
   return 0;
}
```

```python
%%shell
flex program.l
gcc lex.yy.c -o a.out -lfl
./a.out < program.txt
```
